```python
#Student Marks 2D array
marks= [
    [78,99,87],
    [92,84,78],
    [78,88,93],
    [90,83,76],
    [78,65,50]
]
#Maximum Marks
maximum=max(max(row) for row in marks)
print("Maximum Marks=", maximum)

#Minimum Marks
minimum=min(min(row) for row in marks)
print("Minimum Marks=", minimum)

#Average Marks
total=sum(sum(row) for row in marks)
average=total/15
print("Average Marks=", average)

#Student id (0 to 5) who score maximum marks in sub 1
student_id=max(range(5), key=lambda i: marks[i][0])
print("Student Id with maximum marks in Subject 1=", student_id)

#Maximum & Average Marks subject wise
for j in range(3):
    maximum=max(marks[i][j] for i in range(5))
    total=sum(marks[i][j] for i in range(5))
    average=total/5

    print("\nSubject", j+1)
    print("Maximum Marks=", maximum)
    print("Average Marks=", average)


#add 10 marks for all students who score less than 50 in sub 1.
for i in range (5):
    if marks [i][0]<50:
        marks[i][0]+=10
print("\nMarks after adding 10:")
for row in marks:
    print(row)

#No. of studennts score more than 80 in sub 2.
count =0
for i in range(5):
    if marks[i][1]>80:
        count +=1
print("\nStudent scoring more than 80 in sub 2:", count)

#Minimum marks of student 2
minimum_student2=min(marks[2])
print("\nMinimum mark of student 2:",minimum_student2)

#Maximum marks of student 4
maximum_student2=max(marks[3])
print("\nMaximum mark of student 4:",maximum_student2)
```
